## Experiment 1: Disaster-Related News Retrieval

The objective of this experiment is to evaluate how GDELT can be used as a source of news data for disaster-related research.

```python
import requests
import pandas as pd
```

### Defining the Disaster Queries

The following queries are used to retrieve news articles related to major types of disasters in India.

```python
disaster_queries = {
    "Flood": "flood India",
    "Earthquake": "earthquake India",
    "Cyclone": "cyclone India",
    "Wildfire": "wildfire India"
}

disaster_queries
```

### Retrieving Flood-Related News

The first experiment focuses on flood-related news in India.

The GDELT DOC 2.0 API is queried for recent articles matching the search term `flood India`.

```python
import time

time.sleep(10)

url = "https://api.gdeltproject.org/api/v2/doc/doc"

params = {
    "query": disaster_queries["Flood"],
    "mode": "artlist",
    "format": "json",
    "maxrecords": 20,
    "timespan": "7d"
}

response = requests.get(url, params=params, timeout=30)

print("Status Code:", response.status_code)
```

```python
if response.status_code == 200:
    data = response.json()
    print("Number of articles returned:", len(data.get("articles", [])))
else:
    print("Request failed.")
    print("Status Code:", response.status_code)
    print("Response:", response.text[:500])
```

## Observation: API Rate Limiting

During the experiment, the GDELT DOC 2.0 API returned HTTP status code 429
(Too Many Requests). The server response indicated that requests should be
limited to one request every 5 seconds.

This shows that GDELT applies request-rate limitations to API users.
Therefore, API requests need to be controlled carefully when conducting
programmatic data collection.

```python
# Filtering for India-related flood events
india_records = df[
    (df["ActionGeo_CountryCode"] == "IN") |
    (df["Actor1CountryCode"] == "IND") |
    (df["Actor2CountryCode"] == "IND")
].copy()

print("India-associated records:", len(india_records))
print("Percentage of dataset:", round(len(india_records) / len(df) * 100, 2), "%")
```

```python
disaster_keywords = [
    "flood",
    "earthquake",
    "cyclone",
    "storm",
    "wildfire",
    "drought",
    "landslide",
    "tsunami"
]

india_disaster_records = india_records[
    india_records["SOURCEURL"]
    .fillna("")
    .str.lower()
    .str.contains("|".join(disaster_keywords), regex=True)
].copy()

print("India-associated disaster-related records:",
      len(india_disaster_records))

print("\nDisaster keyword matches:")
for keyword in disaster_keywords:
    count = india_records["SOURCEURL"].fillna("").str.lower().str.contains(
        keyword, regex=False
    ).sum()
    print(f"{keyword}: {count}")
```

## Final Findings

The GDELT Event dataset was successfully explored using a downloadable
daily export containing 116,212 event records.

An India-associated subset was created using actor-country and
action-geography fields, resulting in 6,952 records. Disaster-related
candidates were then identified using flood-related keywords in source
URLs.

The initial flood filter produced 34 event records corresponding to
8 unique source articles. This demonstrated that multiple GDELT event
records can originate from the same news article. Therefore, source URLs
were used to obtain an article-level view.

After excluding candidates whose source URLs explicitly referenced
Nepal, 2 potential India-focused flood articles remained.

For these two candidate articles, the average number of mentions was
6, the average number of associated articles was 6, and the average
GDELT tone score was approximately -4.79.

These results should be considered exploratory rather than a validated
India flood dataset. The filtering approach relied on geographic
association and URL keywords, and the GDELT Event dataset does not
contain the full article text. Country association can also arise from
actors or locations mentioned in an article rather than from the
location of the disaster itself.

## Practical Limitations

1. The GDELT DOC API returned HTTP 429 (Too Many Requests) during
   programmatic experimentation, demonstrating the importance of
   rate-limit handling.

2. The downloadable Event dataset provided a practical alternative for
   structured event analysis.

3. Event-level records are not equivalent to individual news articles;
   multiple event records may originate from the same source URL.

4. Simple keyword filtering can produce false positives or geographically
   ambiguous results.

5. The Event dataset provides structured event metadata and source URLs,
   but not the complete article text.

## Research Relevance

GDELT can provide a useful news and event-data modality for a
multimodal disaster-analysis research pipeline. Its structured fields
can support analysis of event frequency, geographic association,
actors, mentions, and media tone.

For a future multimodal system, GDELT data could be combined with
social-media signals and satellite/geospatial observations to study
disasters from complementary perspectives.

This experiment therefore established the practical workflow for
acquiring, structuring, filtering, deduplicating, and critically
evaluating GDELT disaster-related data.
